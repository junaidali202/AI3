```python
import random

# List of 5 predefined words
words = ["python", "computer", "mobile", "school", "bank"]

# Randomly select one word
word = random.choice(words)

# Store correctly guessed letters
guessed_letters = []

# Maximum incorrect guesses
max_wrong_guesses = 6
wrong_guesses = 0

print("================================")
print("       HUMAN GUESSING GAME")
print("================================")
print("Guess the word one letter at a time!")
print("You have 6 incorrect guesses.")

# Game loop
while wrong_guesses < max_wrong_guesses:

    # Display the word with hidden letters
    display_word = ""

    for letter in word:
        if letter in guessed_letters:
            display_word += letter + " "
        else:
            display_word += "_ "

    print("\nWord:", display_word)
    print("Wrong guesses:", wrong_guesses, "/", max_wrong_guesses)

    # Check if the whole word has been guessed
    if "_" not in display_word:
        print("\nCongratulations! ")
        print("You guessed the word:", word)
        break

    # Ask the player for one letter
    guess = input("Guess one letter: ").lower()

    # Check input
    if len(guess) != 1:
        print("Please enter only ONE letter.")
        continue

    if not guess.isalpha():
        print("Please enter a letter.")
        continue

    # Check if letter was already guessed
    if guess in guessed_letters:
        print("You already guessed that letter.")
        continue

    # Add guess to list
    guessed_letters.append(guess)

    # Check whether the letter is in the word
    if guess in word:
        print("Correct guess! ")
    else:
        wrong_guesses += 1
        print("Wrong guess! ")

# If player uses all 6 wrong guesses
if wrong_guesses == max_wrong_guesses:
    print("\nGame Over! ")
    print("The correct word was:", word)
```
